# Introduction — the shear channel

So far every result has rested on one object: the first-order (scalar:  $\delta E^{(v)}$, not a slope: $dE/dk$) energy shift a valley feels under strain, i.e., how far valley $v$ moves up or down when the crystal is deformed. Herring and Vogt (1956), that came out of the Hamiltonian perturbation, wrote it as:

$$ H_{\mathrm{HV}}^{(v)} = \delta E^{(v)} = \Xi_d\,\mathrm{Tr}(\varepsilon) + \Xi_u\,\hat{m}^{(v)\mathsf{T}}\varepsilon\,\hat{m}^{(v)}$$

and the two terms carried the cases we already handled: the trace $\mathrm{Tr}(\varepsilon)$ is the hydrostatic channel (all valleys shift together, the gap moves, carrier density changes), and the sandwich $\hat m^{\mathsf T}\varepsilon\hat m$ is the uniaxial channel (valleys along different axes shift by different amounts, they repopulate, and $\pi_{11}, \pi_{12}$ come out).

Why exactly two terms, and not more, has a clean geometrical reason. The energy shift is a scalar — one number pinned to a valley — so it cannot change when we apply a symmetry operation that leaves that valley untouched. Take the $z$-valley, the cigar along $[001]$: it is unchanged by a $90^\circ$ rotation about $z$. Under that rotation $\varepsilon_{xy}\to-\varepsilon_{xy}$ and $(\varepsilon_{xx}-\varepsilon_{yy})\to-(\varepsilon_{xx}-\varepsilon_{yy})$ — they flip sign, but $\delta E$ cannot, because it is the same valley. So their coefficients must be zero. The only combinations that survive the rotation unchanged are $\varepsilon_{zz}$ and $\varepsilon_{xx}+\varepsilon_{yy}$: two of them, hence two deformation potentials, and the Herring–Vogt form is *complete* at first order rather than a lucky guess. This already contains the punchline — $\varepsilon_{xy}$, the trigonal-shear component, is one of the sign-flipping combinations, so it contributes **nothing** to the energy of the $z$-valley, and the same holds for all six. See the proof in the next cell.




*(The 'sandwich' shape itself is just bookkeeping: to turn a rank-2 tensor like $\varepsilon$ into a scalar you contract it from both sides, hence $\hat m^{\mathsf T}\varepsilon\hat m$.)*

We have also seen how strain reshapes the crystal's symmetry. Silicon is space group $227$, $Fd\bar 3m$:

```
F  = face-centred lattice           → fcc
d  = diamond glide                  → the nonsymmorphic ingredient
3̄  = 3-fold + inversion axis        → carries the inversion centre
m  = mirror
```

with the 48 operations of the point group $O_h$:

$$E,\ 8C_3,\ 3C_2,\ 6C_2,\ 6C_4,\ i,\ 8S_6,\ 3\sigma_h,\ 6\sigma_d,\ 6S_4.$$

A strain can only *remove* operations — the surviving group is the subset that still maps the deformed crystal onto itself,

$$G(\varepsilon)=\{\,g\in O_h : g\,\varepsilon\,g^{\mathsf T}=\varepsilon\,\}.$$

And because the two ends $\pm\mathbf k$ of each cigar are locked together by time reversal, we really only track three valley-pairs, one along each of $x,y,z$. A splitting is allowed only if some surviving operation fails to shuffle all three into one another:

- **unstrained** — $O_h$ (48). The $C_4$'s about the cube axes send $x\to y\to z$, tying all three pairs together → forced degenerate, six equally filled cigars.
- **uniaxial [100]** — $O_h\to D_{4h}$ (16). The strain keeps a $4$-fold axis but kills the rotations that sent the strain axis into the other two. Two families remain — the axial pair and the four transverse valleys — nothing forces them equal → $2+4$ split → repopulation → $\pi_{11},\pi_{12}$.
- **trigonal shear** — $O_h\to D_{3d}$ (12). The surviving rotation is the $3$-fold about the body diagonal $[111]$, and it cycles $x\to y\to z\to x$. That *single* operation ties all three pairs into **one** family → forced to identical energy → identical occupation → no repopulation → $\pi_{44}^{\text{(repop)}}=0$. One rotation settles it; we never need the other eleven.

The twelve operations of $D_{3d}$, spelled out — $E,\ 2C_3,\ 3C_2,\ i,\ 2S_6,\ 3\sigma_d$:

- $2C_3$: rotations by $\pm120^\circ$ about $[111]$ (the one doing the work above),
- $3C_2$: $180^\circ$ about the three $\langle1\bar10\rangle$ axes lying perpendicular to $[111]$,
- $i$: inversion, sitting at the bond midpoint,
- $2S_6$: $\pm60^\circ$ about $[111]$ then reflection in the plane $\perp[111]$,
- $3\sigma_d$: reflections in the three $\{110\}$ planes that *contain* $[111]$.


Group theory tells us the six valleys stay degenerate, but not what the shear actually *does* to them. So put the shear straight into the sandwich and ask what a cigar pointing in a general direction $\hat n$ would feel. Trigonal shear means all off-diagonals equal, diagonals zero, $\varepsilon_{xy}=\varepsilon_{yz}=\varepsilon_{zx}=\varepsilon$:

$$\hat n^{\mathsf T}\varepsilon\,\hat n = 2\varepsilon\,(n_xn_y+n_yn_z+n_zn_x)=\varepsilon\,(3\cos^2\theta-1),$$

where $\theta$ is the angle between $\hat n$ and $[111]$ (using $n_x+n_y+n_z=\sqrt3\cos\theta$). So the shear's energy-shift field is a single lobe about the body diagonal: it lifts a cigar pointing near $[111]$ by up to $+2\varepsilon\Xi_u$, pushes one lying in the $(111)$ plane down by $-\varepsilon\Xi_u$, and gives **exactly zero** on the cone where $3\cos^2\theta=1$, i.e.

$$\theta_0=\arccos\tfrac{1}{\sqrt3}=54.74^\circ.$$

And the whole thing in one line: **the cube axes make exactly $54.74^\circ$ with the body diagonal.** Every one of the six $\Delta$-valleys sits *on the null cone*. The shear does not fail to reach them — it reaches all six with identical weight, and the weight is zero, because the stretch it wants to give along $[111]$ and the squeeze in the $(111)$ plane cancel at that magic angle. That is the geometric answer to whether there is a shear effect: there *is* a real one on generic directions — an up/down lobe about $[111]$ — but the six places we care about all lie exactly where it vanishes.

It is tempting to read $\pi_{44}=0$ as a symmetry *prohibition*. It is not. Under $D_{3d}$ the conductivity is allowed to be uniaxial about $[111]$, and written back in cube axes that already carries a nonzero $\sigma_{xy}$ — so a finite $\pi_{44}$ is symmetry-*allowed* (that is exactly why the cubic tensor needs three numbers $\pi_{11},\pi_{12},\pi_{44}$, not two). What our theorem actually kills is narrower: the **repopulation** mechanism — the one that produced $\pi_{11},\pi_{12}$ — is switched off, because the six valleys never split. And that rests on three assumptions:

1. rigid bands (energies shift, shapes fixed),
2. strain-independent valley masses,
3. a single constant $\tau$.

With all three, $\pi_{44}=0$. Note carefully what this rules out and what it doesn't: a fancier Boltzmann solver fed the *same* rigid bands returns the same zero — the extra physics a Monte-Carlo BTE adds lives in the collision integral, but the only strain information we handed it sits in valley energies that did not move. So a better *solver* cannot revive $\pi_{44}$; new *input* can. Drop assumption 2 or 3 and it comes back — and experiment says it does: n-Si has $\pi_{44}\approx-13.6$ against $\pi_{11}\approx-102$ (in $10^{-11}\,\text{Pa}^{-1}$). Small, but not zero, and that survivor is the subject of the rest.

Loosening the assumptions one at a time turns up two doors, and one is already open to us.

The **mass door** (drop assumption 2) is reachable with the machinery we already have. Symmetry fixed the valley *energies* equal; it said nothing about their *shapes*. Under $\varepsilon_{xy}$ the $z$-valley loses the $4$-fold symmetry that had forced its two transverse masses equal and its cross-term $m_{xy}$ to vanish, so the transverse ellipse is now free to go lopsided, with axes rotated onto $[110]$ and $[1\bar10]$. And $m_{xy}$ must be *odd* in $\varepsilon_{xy}$ (flipping the shear swaps $[110]\leftrightarrow[1\bar10]$), so it should turn on **linearly** — a first-order distortion. Since $\sigma=\sum_v n_v e^2\tau\,(M^{(v)})^{-1}$, a tilted mass tensor feeds $\sigma_{xy}$ directly, even with all six occupations equal and $\tau$ fixed. This is the physics behind the $[110]$ mobility boost in strained nMOS (Sun, Thompson & Nishida, already on the shelf). Crucially, this door opens with VASP alone: impose a finite $\varepsilon_{xy}$, compute the bands, fit the mass tensor at the valley minima, and read off whether $m_{xy}$ really turns on linearly and how large it is. No EPW.

Worth flagging why it is subtle: $\varepsilon_{xy}$ splits the degeneracy the $z$-pair carries at $X$, and the valley minimum — sitting only ~15% of the way back from $X$ toward $\Gamma$ — feels a *small* energy denominator. The *energy* of the minimum only drops at **second** order in $\varepsilon$ (consistent with our linear null), while the *shape* responds at first order and can be strong and nonlinear. Which of the energy-null and the mass-signal dominates at each order is precisely what the mass fit will tell us.

The **scattering door** (drop assumption 3) is the one that genuinely needs EPW. Even with rigid bands and fixed masses, a strain- and valley-dependent $\tau(\varepsilon)$ produces $\pi_{44}$; it lives entirely in the electron–phonon coupling $g(\mathbf k,\mathbf q,\varepsilon)$ and is invisible to any constant-$\tau$ treatment. This is the beyond-RTA stretch goal — and it is the *only* part of $\pi_{44}$ that actually requires EPW. So the earlier framing, "the whole $\pi_{44}$ channel is gated behind EPW," was too strict: one of its two sources is a band calculation we can run now.

One crystal note, since shear is where it bites: a Si atom's site symmetry is only $T_d$, not the full $O_h$ — the inversion centre sits at the bond midpoint, between the two sublattices, not on an atom. Shear is the one channel that slides the two sublattices past each other, so the internal (Kleinman) displacement rides along with the macroscopic strain and must be *relaxed*, not frozen, when we compute the shear bands.

The sharpest way to show this channel-decomposition is real, and not an artefact of one material, is to run the same shear on Ge, whose conduction valleys sit at the $L$ points along the $\langle111\rangle$ diagonals. Into the sandwich again: the valley along $[111]$ points straight up the lobe and shifts by $+2\varepsilon\Xi_u$, while the other three lie at $\arccos(-\tfrac13)=70.5^\circ$ from it and each shift by $-\tfrac23\varepsilon\Xi_u$ — a $1+3$ split, traceless as required. Ge's valleys are **off** the null cone, so shear splits them and $\pi_{44}$ is large already at constant $\tau$. And the mirror holds the other way: under $[100]$ strain all four $L$-valleys make the same angle with $[100]$, shift identically, and do *not* split. So:

- **Silicon** — blind to trigonal shear (valleys on the null cone), strong under $[100]$.
- **Germanium** — blind to $[100]$ (valleys equally tilted to it), strong under trigonal shear.

Smith (1954) says exactly this: n-Ge has $\pi_{11}\approx\pi_{12}\approx-5$ with $\pi_{44}\approx-138$ — the reflection of n-Si. Setting our Si *null* beside the Ge *signal* turns "we computed zero" into a two-sided statement: which strain a crystal responds to is decided by where its valleys sit relative to the strain's geometry. *(Reminder for later: the compliance conversion has to be rebuilt for Ge first — its factors are nowhere near Si's near-unity values.)*

We now go through the first door — the shape of the ellipsoids under $\varepsilon_{xy}$ — with nothing more than band structures and mass fits, and see how far the energy derivatives carry us before we have to open the electron–phonon interaction for the second.

### Setup
The z-valley shift is linear in strain, i.e. a constant symmetric tensor $D^{(z)}$ (the deformation-potential tensor) contracted with $\varepsilon$:
$$\delta E^{(z)}(\varepsilon) = D^{(z)}:\varepsilon = \sum_{ij}D^{(z)}_{ij}\varepsilon_{ij}.$$

### Theorem
If $R$ fixes the z-valley (i.e. $R$ lies in its little group; for Si that is $C_{4v}$, so in particular $R=C_4^z$), then $\delta E^{(z)}$ can depend on only **two** independent combinations of strain,
$$\delta E^{(z)} = \Xi_d\,\mathrm{Tr}(\varepsilon)+\Xi_u\,\varepsilon_{zz},\qquad \frac{\partial\,\delta E^{(z)}}{\partial\varepsilon_{xy}}=0.$$

### Proof
The only physical input: an operation that maps the z-valley onto itself cannot change the z-valley's energy. Such an operation sends the strain $\varepsilon \mapsto R\varepsilon R^{\mathsf T}$, so
$$
\begin{align*}
D^{(z)}:\varepsilon &= D^{(z)}:(R\varepsilon R^{\mathsf T}) &&\forall\,\varepsilon\\
&= (R^{\mathsf T} D^{(z)} R):\varepsilon &&\forall\,\varepsilon\\
\Rightarrow\quad R^{\mathsf T} D^{(z)} R &= D^{(z)}.
\end{align*}
$$
Now put in $R=C_4^z$ (the $90^\circ$ turn about $z$). Componentwise, $R^{\mathsf T}D^{(z)}R=D^{(z)}$ reads
$$
\begin{align*}
D_{xx}&=D_{yy},\\
D_{xy}&=-D_{xy}\;\Rightarrow\;D_{xy}=0,\\
(D_{xz},D_{yz})&=(-D_{yz},D_{xz})\;\Rightarrow\;D_{xz}=D_{yz}=0,
\end{align*}
$$
so the tensor is forced diagonal with equal transverse entries:
$$
D^{(z)}=\begin{pmatrix}d_\perp&0&0\\0&d_\perp&0\\0&0&d_\parallel\end{pmatrix}
= d_\perp\,\mathbb{1}+(d_\parallel-d_\perp)\,\hat z\hat z^{\mathsf T}.
$$
Contract back with $\varepsilon$:
$$
\begin{align*}
\delta E^{(z)}
&= d_\perp\,(\mathbb{1}:\varepsilon)+(d_\parallel-d_\perp)\,(\hat z^{\mathsf T}\varepsilon\,\hat z)\\
&= d_\perp\,\mathrm{Tr}(\varepsilon)+(d_\parallel-d_\perp)\,\varepsilon_{zz}\\
&\equiv \Xi_d\,\mathrm{Tr}(\varepsilon)+\Xi_u\,\hat m^{(z)\mathsf T}\varepsilon\,\hat m^{(z)}\quad\blacksquare
\end{align*}
$$
with $\Xi_d=d_\perp$, $\Xi_u=d_\parallel-d_\perp$, $\hat m^{(z)}=\hat z$. This is the Herring–Vogt form, and it is *complete*: two constants, no room for a third. Because $D^{(z)}$ carries no $xy$ entry, $\varepsilon_{xy}$ drops out — a pure trigonal shear shifts the z-valley by zero at first order. The other five valleys follow by the $O_h$ rotations that permute them onto z.

### Reading it on the strain side
The line $D_{xy}=-D_{xy}$ is the crux, and it has a plain meaning: under $C_4^z$ the component $\varepsilon_{xy}$ flips sign (as do $\varepsilon_{xx}-\varepsilon_{yy}$ and, mixing, $\varepsilon_{xz},\varepsilon_{yz}$), whereas $\mathrm{Tr}(\varepsilon)$ and $\varepsilon_{zz}$ are left alone. A quantity forced to equal its own negative is zero, so the z-valley energy can only ride on the two components the rotation leaves untouched. Hydrostatic and $\varepsilon_{zz}$ get through; the shear does not — which is the same statement as $D_{xy}=0$, now read off the strain rather than the tensor.

## Phonon dispersion analysis for strained Si

**The electron–phonon scattering rate :**

$$\frac{1}{\tau_{n\mathbf{k}}} = \frac{2\pi}{\hbar}\sum_{m,\nu}\int\frac{d\mathbf{q}}{\Omega_{BZ}}\;\big|g_{mn\nu}(\mathbf{k},\mathbf{q})\big|^2\;\Big[(n_{\mathbf{q}\nu}+1-f_{m\mathbf{k+q}})\,\delta(\varepsilon_{n\mathbf{k}}-\varepsilon_{m\mathbf{k+q}}-\hbar\omega_{\mathbf{q}\nu}) + (n_{\mathbf{q}\nu}+f_{m\mathbf{k+q}})\,\delta(\varepsilon_{n\mathbf{k}}-\varepsilon_{m\mathbf{k+q}}+\hbar\omega_{\mathbf{q}\nu})\Big]$$


**From phonon calculation — 🟢 we have these:**
- **ω_{qν}** — the phonon frequency of branch ν at wavevector q. *This is literally your dispersion curve.* It appears twice: inside the δ-functions (energy conservation — the electron's energy change must equal ±one phonon quantum ℏω) and inside n_{qν} below.
- **n_{qν}** — the phonon occupation (Bose–Einstein): n = 1/(e^{ℏω/k_BT} − 1). Built *from* your ω. The two terms (n+1) and (n) are phonon **emission** and **absorption** — the "+1" is spontaneous emission.
- The phonon **eigenvectors** (polarizations) — hidden inside g (see below), because g depends on *which direction* the atoms move.

**From plain DFT  — 🟢 we have these:**
- **ε_{nk}, ε_{mk+q}** — electron band energies before and after scattering. Your band structure. Inside the δ-functions.
- **f_{mk+q}** — electron occupation (Fermi–Dirac) of the final state. From your band structure + Fermi level.

**From EPW  — 🔴 the missing piece:**
- **g_{mnν}(k,q)** — the electron–phonon coupling matrix element. This is the *only* term you don't have. It measures how strongly an electron in state (n,k) scatters to (m,k+q) by emitting/absorbing phonon (q,ν). It's the derivative of the electronic potential w.r.t. the phonon displacement, projected between Bloch states — and computing it *requires your phonon eigenvectors as input* (the atoms have to move *somehow*, and the eigenvector says how).

**So the flow, made concrete:**

$$\underbrace{\varepsilon_{n\mathbf{k}}}_{\text{DFT ✅}} \;,\; \underbrace{\omega_{\mathbf{q}\nu}\,,\,\text{eigenvectors}}_{\text{your phonons ✅}} \;\longrightarrow\; \underbrace{g_{mn\nu}(\mathbf{k},\mathbf{q})}_{\text{EPW 🔴}} \;\longrightarrow\; \underbrace{\frac{1}{\tau_{n\mathbf{k}}}}_{\text{scattering rate}} \;\longrightarrow\; \tau \;\longrightarrow\; \mu = \frac{e\tau}{m^*} \;\longrightarrow\; \sigma$$

**Where your results go, in one sentence:** your ω(q,ν) and phonon eigenvectors feed into *two* places — directly into the energy-conservation δ-functions and Bose factors of this formula, and *indirectly* into g (which needs the eigenvectors to know how atoms move). EPW computes g using your phonons + the DFT electrons, then this integral gives 1/τ, then τ → μ → σ. Your phonon calculation is the vibrational half of the input; EPW is the machine that combines it with the electrons into the coupling; the formula assembles everything into a scattering rate.

**Now the part that makes this *your* thesis and not a generic mobility calculation — the strain axis:**

For π₄₄, every strain-dependent quantity gets an ε:

$$\frac{1}{\tau_{n\mathbf{k}}(\varepsilon)} \sim \sum \big|g_{mn\nu}(\mathbf{k},\mathbf{q};\boldsymbol{\varepsilon})\big|^2\,\delta(\ldots\pm\hbar\omega_{\mathbf{q}\nu}(\boldsymbol{\varepsilon}))\ldots$$

Your theorem said: under constant τ, π₄₄ = 0 because trigonal shear doesn't split valleys — so the δ-functions and band energies don't change to first order. **The entire π₄₄ signal therefore lives in g(k,q;ε) and ω(q;ε) — the strain-dependence of the coupling and the phonons.** That's why:

- **ω(q;ε)** — this notebook rerun on sheared cells — is one of the two strain-dependent ingredients.
- **g(k,q;ε)** — EPW on sheared cells — is the other, and the harder one.

For each of the strains we already have the `POSCAR` files from strain sweeps, and for each of that strain sweeps, we run the `DFT` finite differences + `phonopy` calculations, to get the $w(q,\varepsilon)$

## Input files for the strained-phonon calculation 

### The two INCARs

The calculation has two stages, so two INCARs:
- **`INCAR.relax`** — relax the ions inside the strained cell to equilibrium (find the Kleinman internal displacement).
- **`INCAR.force`** — compute forces on each *displaced* supercell (the finite-difference step that gives force constants).

### Shared tags 

| tag | value | why |
|---|---|---|
| `ISTART` | 0 | Start fresh — no wavefunction from a previous run to read. |
| `ICHARG` | 2 | Build the initial charge density from a superposition of atomic densities (the standard self-consistent start). |
| `ENCUT` | 320 | Plane-wave kinetic-energy cutoff (eV). Sets basis-set size/accuracy. **Must be identical across every run in the sweep** — comparing energies/forces at different cutoffs is meaningless. 320 is a reasonable Si value; ideally converge-test it once. |
| `ISMEAR` | 0 | Gaussian smearing of the electronic occupations near the Fermi level. `0` (Gaussian) is the safe choice for semiconductors/insulators like Si. (Metals would use `ISMEAR=1` or `2`; the tetrahedron method `-5` is best for final DOS but not for forces.) |
| `SIGMA` | 0.05 | Width (eV) of that smearing. Small, because Si has a gap — you don't want to smear across it. |
| `ISYM` | 0 | **Symmetry off.** Critical for strained cells: shear breaks cubic symmetry, and letting VASP impose the wrong symmetry corrupts the forces. `0` makes VASP treat the reduced symmetry honestly. |
| `EDIFF` | 1E-8 | Convergence threshold for the **electronic** self-consistency loop (energy change between SCF steps, in eV). **Very tight** here because force constants are *second* derivatives — tiny SCF noise gets amplified into large phonon errors. This is much tighter than a normal SCF (1E-6); phonons demand it. |
| `GGA` | PE | The exchange-correlation functional: PBE (Perdew–Burke–Ernzerhof). Matches your whole pipeline. |
| `LREAL` | .FALSE. | Evaluate projectors in reciprocal space (more accurate than real-space `.TRUE.`). For a small cell, always `.FALSE.` — accuracy matters for forces. |

### Tags in `INCAR.relax` (the relaxation)

| tag | value | why |
|---|---|---|
| `NSW` | 50 | Max number of ionic relaxation steps. The relaxation must converge *within* this budget. |
| `IBRION` | 2 | The ionic-relaxation algorithm (conjugate gradient). Moves atoms downhill in energy. |
| `ISIF` | 2 | **Relax ions only; hold cell fixed.** This is the one that preserves your imposed strain — `ISIF=3` would relax the cell shape and erase the shear. Essential. |
| `EDIFFG` | −1E-4 | Ionic convergence: stop when all forces are below 1E-4 eV/Å (the minus sign means "force criterion," not energy). Tight, because leftover forces contaminate the phonons. |

### Tags in `INCAR.force` (the finite-displacement forces)

| tag | value | why |
|---|---|---|
| `IBRION` | −1 | **Do not move the ions.** Compute forces at the given (displaced) geometry and stop. This is exactly what finite-displacement phonons need: forces *at* the displaced configuration. |
| `NSW` | 1 | One "step" = a single static force evaluation, no relaxation. |

(No `ISIF`/`EDIFFG` here — we're not relaxing, just reading forces.)

### KPOINTS files

KPOINTS samples **reciprocal space** (the electronic Brillouin-zone integration).

- **`KPOINTS.prim` (8×8×8)** — for the **primitive cell** (relaxation stage). Small cell → large Brillouin zone, needs a dense k-mesh.
- **`KPOINTS.sc` (4×4×4)** — for the **2×2×2 supercell** (force stage). The supercell is 8× bigger in real space, its Brillouin zone is 8× *smaller* in reciprocal space, a coarser mesh samples it equivalently. Halving the grid (8 to 4) roughly matches the k-point *density* between the two cells. Using the dense 8×8×8 on the supercell would waste compute; using a coarse grid on the primitive cell would under-converge. **The rule: k-mesh density should be consistent, so bigger cells get coarser grids.**

Both are Γ-centered (`Gamma`), which is safest for phonons (includes the zone center exactly).

### `band.conf` 

It tells phonopy, *after* the forces are computed:
- `DIM = 2 2 2` — the supercell size used (so it knows how the force constants tile).
- `BAND = ...` — the **phonon q-path** (Γ→X→K→Γ→L) in reciprocal coordinates, along which to interpolate and diagonalize D(q).
- `BAND_LABELS`, `BAND_POINTS` — cosmetic (tick labels) and resolution (101 points per segment).

So KPOINTS.prim/sc sample the *electronic* BZ for VASP; band.conf specifies the *phonon* q-path for phonopy. Two different reciprocal-space samplings for two different purposes.


````python
strs = ['-0.0020', '-0.0050', '-0.0100', '+0.0000', '+0.0020', '+0.0050', '+0.0100']
poscar_d = "./eps"
output_d = "./eps_phonon"
RESET_DIR = True

# consistent DFT 
INCAR_scf = """ISTART=0
ICHARG=2
ENCUT=320
ISMEAR=0
SIGMA=0.05
ISYM=0
EDIFF=1E-8
EDIFFG=-1E-4
NSW=50
IBRION=2
ISIF=2
GGA=PE
LREAL=.FALSE.
"""

# Force constant 
INCAR_force = """ISTART=0
ICHARG=2
ENCUT=320
ISMEAR=0
SIGMA=0.05
ISYM=0
EDIFF=1E-8
IBRION=-1
NSW=1
GGA=PE
LREAL=.FALSE.
"""

KPOINTS_PRIM = "auto\n0\nGamma\n8 8 8\n0 0 0\n"    # primitive cell (relax)
KPOINTS_SC   = "auto\n0\nGamma\n4 4 4\n0 0 0\n"    # 2x2x2 supercell (forces)


# phonopy band path: Gamma-X-K-Gamma-L
BAND_CONF = """ATOM_NAME = Si
DIM = 2 2 2
BAND = 0.0 0.0 0.0  0.5 0.0 0.5  0.375 0.375 0.75  0.0 0.0 0.0  0.5 0.5 0.5
BAND_LABELS = \\Gamma X K \\Gamma L
BAND_POINTS = 101
"""


for e in strs:
    if RESET_DIR:
        for f in glob.glob(f'{output_d}/eps_{e}/*'):
            os.remove(f)

    os.makedirs(f'{output_d}/eps_{e}', exist_ok=True)
    
    # read source POSCAR
    content = open(f'{poscar_d}_{e}/POSCAR').read()
    content2 = open(f'{poscar_d}_{e}/POTCAR').read()
    # Copying existing POSCARs to each directory
    open(f'{output_d}/eps_{e}/POSCAR','w').write(content)
    open(f'{output_d}/eps_{e}/POTCAR','w').write(content2)
    open(f'{output_d}/eps_{e}/INCAR.relax','w').write(INCAR_scf)
    open(f'{output_d}/eps_{e}/INCAR.force','w').write(INCAR_force)
    open(f'{output_d}/eps_{e}/KPOINTS.prim','w').write(KPOINTS_PRIM)
    open(f'{output_d}/eps_{e}/KPOINTS.sc','w').write(KPOINTS_SC)
    open(f'{output_d}/eps_{e}/band.conf','w').write(BAND_CONF)

````

##  Step 1:  Relaxation
The bellow cell runs VASP and relaxes the geometry and copies it into POSCAR.relax. This cell means convergence of `Potential energy` and the forces should be driven to zero.

````python
#strs = ['-0.0020', '-0.0050', '-0.0100', '+0.0000', '+0.0020', '+0.0050', '+0.0100']
strs = ['+0.0000']

files = ['POSCAR', 'POTCAR', 'INCAR.relax', 'INCAR.force', 
         'KPOINTS.prim', 'KPOINTS.sc', 'band.conf']


# upload the input dirs
for e in strs:
    cmd = (
    f"{SCP} "
    + " ".join(f"{output_d}/eps_{e}/{f}" for f in files)
    + f" {USER}@{HOST}:{REMOTE}/eps_{e}/"
    )
    run(cmd)
    run(f"{SSH} 'mkdir -p {REMOTE}/eps_{e}'")
    
st1 = """source /etc/profile.d/modules.sh 2>/dev/null || true
module load intel-oneapi-mkl

for d in eps_*/; do
    ( cd "$d" || exit
      cp INCAR.relax INCAR
      cp KPOINTS.prim KPOINTS
      echo "=== $d relaxing ==="
      vasp > log.relax 2>&1
      cp CONTCAR POSCAR            # save relaxed cell for the phonon stage
      grep -q "reached required accuracy" OUTCAR && echo "  OK" || echo "  NOT converged"
    )

done
"""
# The sweeps are already done in script
open(f'{output_d}/s1.sh', 'w').write(st1)
run(f"{SCP} {output_d}/s1.sh {USER}@{HOST}:{REMOTE}/")
run(f"{SSH} 'cd {REMOTE} && bash --login s1.sh'")
````
